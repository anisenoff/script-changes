```python
import json
from pathlib import Path
from statistics import mean
```

```python
out_file = Path(f"../../../../data/httparchive/summary/basic_overlap.json").resolve()
with open(out_file) as json_file:
    data = json.load(json_file)
    
    
crawl_identifiers = []
for month in range(11,13):
    crawl_identifiers.append(f"2024-{month}-1")
for month in range(1,13):
    crawl_identifiers.append(f"2025-{month}-1")
for month in range(1,5):
    crawl_identifiers.append(f"2026-{month}-1")
    
for i in list(data.keys()):
    if not i in crawl_identifiers:
        del data[i]
```

```python







dates = sorted(data.keys())

total_pages = sum(v["total_pages"] for v in data.values())
total_requests = sum(v["total_requests"] for v in data.values())
total_hash = sum(v["requests_w_hash"] for v in data.values())
total_ref = sum(v["requests_w_ref"] for v in data.values())
total_3p = sum(v["requests_3rd_pty"] for v in data.values())
total_3p = sum(v["requests_3rd_pty"] for v in data.values())
total_usable = sum(v["requests_usable"] for v in data.values())

print(f"%measurement period & ", dates[0], dates[-1], "\\\\")
print(f"Number of Crawls & {len(dates):,}", "\\\\")
print(f"Total pages measured & {total_pages:,}", "\\\\")
print(f"Average pages per crawl & {round(mean(v['total_pages'] for v in data.values())):,}", "\\\\")

print(f"Total script requests & {total_requests:,}", "\\\\")
print(f"Average script requests per crawl & {round(mean(v['total_requests'] for v in data.values())):,}", "\\\\")

print(f"Requests with valid hashes & {total_hash:,}", "\\\\")
print(f"Requests with document/referrer context & {total_ref:,}", "\\\\")
print(f"Third-party script requests & {total_3p:,}", "\\\\")
print(f"Requests retained after filtering & {total_usable:,}", "\\\\")

print(f"Percent retained after filtering & ",
      round(100 * total_usable / total_requests, 2), "\\%\\\\")
```
